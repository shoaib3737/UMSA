```python
import moviepy.editor as mp
import os

def video_to_audio(video_path, audio_path):
    """Convert video file to audio file in WAV format."""
    video_files = [f for f in os.listdir(video_path) if f.endswith(('.mp4', '.mov', '.avi'))]

    for video_file in video_files:
        video_full_path = os.path.join(video_path, video_file)
        audio_file_name = os.path.splitext(video_file)[0] + '.wav'  # Change extension to .wav
        audio_full_path = os.path.join(audio_path, audio_file_name)

        try:
            # Convert video to audio
            video = mp.VideoFileClip(video_full_path)
            video.audio.write_audiofile(audio_full_path, codec='pcm_s16le')
            print(f"Successfully converted {video_file} to {audio_file_name}")
        except Exception as e:
            print(f"Error processing {video_file}: {e}")

# Define paths
video_file_path = '/content/drive/MyDrive/work2/videos/videos_37'
audio_file_path = '/content/drive/MyDrive/work2/Training_Data_Phase_2/training_audios37'

# Ensure audio path exists
os.makedirs(audio_file_path, exist_ok=True)

# Convert videos to audio
video_to_audio(video_file_path, audio_file_path)

print("Video to audio conversion completed.")
```

```python
import os
import pandas as pd
import speech_recognition as sr
from pydub import AudioSegment

def get_audio_length(file_path):
    """Return the length of the audio file in seconds."""
    audio = AudioSegment.from_file(file_path)
    return len(audio) / 1000.0  # Convert milliseconds to seconds

def transcribe_audio_speech_recognition(audio_file_path):
    """Transcribe the given local audio file using the speech_recognition library."""
    recognizer = sr.Recognizer()
    with sr.AudioFile(audio_file_path) as source:
        audio_data = recognizer.record(source)
        try:
            transcript = recognizer.recognize_google(audio_data, language="ur-PK")
            return transcript
        except sr.UnknownValueError:
            return None
        except sr.RequestError:
            return None

def process_audios(audio_path, output_csv):
    """Process audio files to transcribe and save to CSV."""
    results = []

    # List all audio files in the directory
    audio_files = [f for f in os.listdir(audio_path) if f.endswith('.wav')]

    for audio_file in audio_files:
        audio_full_path = os.path.join(audio_path, audio_file)

        # Transcribe audio
        transcript = transcribe_audio_speech_recognition(audio_full_path)
        length_in_seconds = get_audio_length(audio_full_path)

        # Calculate word count
        word_count = len(transcript.split()) if transcript else 0
        status = "Transcribed" if transcript else "Not Transcribed"

        # Append result
        results.append({
            'File': audio_file,
            'Transcription': transcript if transcript else '',
            'Status': status,
            'Length': round(length_in_seconds, 2),
            'WordCount': word_count,
            'Senti': 0  # Assuming sentiment analysis is 0 for now
        })
    # Create a DataFrame and save to CSV
    df = pd.DataFrame(results)
    df.to_csv(output_csv, index=False, encoding='utf-8-sig')

# Define paths
audio_file_path = '/content/drive/MyDrive/work2/Training_Data_Phase_2/training_audios37'
output_csv_path = '/content/drive/MyDrive/work2/Training_Data_Phase_2/transcriptions37.csv'

# Process audio files and save to CSV
process_audios(audio_file_path, output_csv_path)

print("Audio transcriptions have been saved to CSV.")
```

```python
import cv2
import os
import pandas as pd

# Function to extract frames at each second
def extract_frames_per_second(video_file_path, output_folder):
    video_name = os.path.basename(video_file_path).split('.')[0]
    cap = cv2.VideoCapture(video_file_path)
    frame_info = []
    frame_rate = cap.get(cv2.CAP_PROP_FPS)
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    duration = int(total_frames / frame_rate)  # Total duration in seconds

    for current_time in range(duration):
        cap.set(cv2.CAP_PROP_POS_MSEC, current_time * 1000)
        ret, frame = cap.read()
        if not ret:
            break
        frame_filename = os.path.join(output_folder, f'{video_name}_frame_{current_time}.jpg')
        cv2.imwrite(frame_filename, frame)
        frame_info.append({'Video Name': video_name, 'Frame Path': frame_filename, 'Time in Seconds': current_time})

    cap.release()
    return frame_info

# Directory containing the videos to process
input_video_directory = '/content/drive/MyDrive/work2/videos/videos_2'
output_frames_base_folder = '/content/drive/MyDrive/work2/Training_Data_Phase_2/Frames'

# Initialize list to store frame information
all_frame_info = []

# List all video files in the input directory
video_files = [f for f in os.listdir(input_video_directory) if f.endswith(('.mp4', '.avi', '.mov'))]

# Process each video file
for video_file in video_files:
    video_file_path = os.path.join(input_video_directory, video_file)
    video_name = os.path.splitext(video_file)[0]

    # Create output folder for each video
    output_folder = os.path.join(output_frames_base_folder, video_name)
    os.makedirs(output_folder, exist_ok=True)

    # Extract frames at each second
    frame_info = extract_frames_per_second(video_file_path, output_folder)

    # Append extracted frame information to the combined list
    all_frame_info.extend(frame_info)

# Save combined frame information to a CSV file
csv_file_path = '/content/drive/MyDrive/work2/Training_Data_Phase_2/All_Frames_Info.csv'
frame_info_df = pd.DataFrame(all_frame_info)
frame_info_df.to_csv(csv_file_path, index=False)

print(f"All frames extracted and saved in respective folders under: {output_frames_base_folder}")
print(f"All frame information saved to {csv_file_path}")
```

```python
import cv2
import os

# Input and output folder paths
Input_Folder = "/content/drive/MyDrive/work2/Training_Data_Phase_2/Frames"
Output_Folder = "/content/drive/MyDrive/work2/Training_Data_Phase_2/Frames37"

# Function to crop black borders
def crop_black_borders(image_path, output_path):
    image = cv2.imread(image_path)

    # Convert the image to grayscale
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    # Threshold to create a binary mask of the non-black areas
    _, binary_mask = cv2.threshold(gray, 1, 255, cv2.THRESH_BINARY)

    # Find contours of the non-black area
    contours, _ = cv2.findContours(binary_mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

    # Get the bounding box of the largest contour
    if contours:
        x, y, w, h = cv2.boundingRect(max(contours, key=cv2.contourArea))
        cropped_image = image[y:y+h, x:x+w]

        # Ensure output directory exists
        os.makedirs(os.path.dirname(output_path), exist_ok=True)

        # Save the cropped image
        cv2.imwrite(output_path, cropped_image)
        print(f"Saved cropped image: {output_path}")
    else:
        print(f"No valid contours found for: {image_path}")

# Walk through all subfolders and files in the input folder
for root, _, files in os.walk(Input_Folder):
    for file in files:
        # Process only image files
        if file.lower().endswith(('.png', '.jpg', '.jpeg')):
            input_image_path = os.path.join(root, file)

            # Create corresponding output path
            relative_path = os.path.relpath(root, Input_Folder)
            output_image_path = os.path.join(Output_Folder, relative_path, file)

            # Crop and save the image
            crop_black_borders(input_image_path, output_image_path)
```

```python
import os
import pandas as pd
import speech_recognition as sr
from pydub import AudioSegment

def get_audio_length(file_path):
    """Return the rounded length of the audio file in seconds."""
    audio = AudioSegment.from_file(file_path)
    return int(len(audio) / 1000)  # Convert milliseconds to seconds and round down

def transcribe_audio_speech_recognition(audio_file_path):
    """Transcribe the given local audio file using the speech_recognition library."""
    recognizer = sr.Recognizer()
    with sr.AudioFile(audio_file_path) as source:
        audio_data = recognizer.record(source)
        try:
            return recognizer.recognize_google(audio_data, language="ur-PK")
        except sr.UnknownValueError:
            print("Google Speech Recognition could not understand the audio")
            return None
        except sr.RequestError as e:
            print(f"Could not request results from Google Speech Recognition service; {e}")
            return None

def read_words_from_file(file_path):
    """Read words from a file and return them as a list."""
    try:
        with open(file_path, 'r', encoding='utf-8') as file:
            return [line.strip().lower() for line in file.readlines()]
    except FileNotFoundError:
        print(f"File not found: {file_path}")
        return []
    except Exception as e:
        print(f"Error reading file {file_path}: {e}")
        return []

def analyze_sentiment(words_in_transcript, negative_words, positive_words, length_in_seconds):
    negative_count, positive_count = 0, 0
    negative_positions, positive_positions = [], []
    for index, word in enumerate(words_in_transcript):
        position = (index + 1) // (len(words_in_transcript) / length_in_seconds)
        if word.lower() in negative_words:
            negative_count += 1
            negative_positions.append(round(position))
        elif word.lower() in positive_words:
            positive_count += 1
            positive_positions.append(round(position))
    return negative_count, positive_count, negative_positions, positive_positions

# Path to the specific audio file
audio_file_path = '/content/drive/MyDrive/work2/Test/Audio/2.wav'
negative_words = read_words_from_file('/content/drive/MyDrive/work2/dictionary/urdu_negative_words.txt')
positive_words = read_words_from_file('/content/drive/MyDrive/work2/dictionary/urdu_positive_words.txt')

results = []

if os.path.exists(audio_file_path):
    audio_file = '2.wav'
    length_in_seconds = get_audio_length(audio_file_path)
    transcript = transcribe_audio_speech_recognition(audio_file_path)

    if transcript:
        words_in_transcript = transcript.split()
        negative_count, positive_count, negative_positions, positive_positions = analyze_sentiment(words_in_transcript, negative_words, positive_words, length_in_seconds)

        results.append({
            'File': audio_file,
            'Transcription': transcript,
            'Status': 'Transcribed',
            'Length': length_in_seconds,
            'WordCount': len(words_in_transcript),
            'negative_count': negative_count,
            'positive_count': positive_count,
            'negative_positions': negative_positions,
            'positive_positions': positive_positions
        })
    else:
        results.append({
            'File': audio_file,
            'Transcription': '',
            'Status': 'Not Transcribed',
            'Length': length_in_seconds,
            'WordCount': 0,
            'negative_count': 0,
            'positive_count': 0,
            'negative_positions': [],
            'positive_positions': []
        })
else:
    print(f"Audio file not found: {audio_file_path}")

# Save results to a CSV file
csv_file_path = '/content/drive/MyDrive/work2/Test/key_positions_result_2wav.csv'
df = pd.DataFrame(results)
df.to_csv(csv_file_path, index=False)
print(f"Results saved to {csv_file_path}")
```

```python
import os
import pandas as pd
import cv2

def extract_frames(video_file_path, times_in_seconds, output_folder):
    """Extract frames at specified times and save them to the output folder."""
    video_name = os.path.basename(video_file_path).split('.')[0]
    cap = cv2.VideoCapture(video_file_path)
    frame_info = []
    extracted_times = set()
    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break
        current_time = int(cap.get(cv2.CAP_PROP_POS_MSEC) // 1000)
        if current_time in times_in_seconds and current_time not in extracted_times:
            frame_filename = os.path.join(output_folder, f'{video_name}_frame_{current_time}.jpg')
            cv2.imwrite(frame_filename, frame)
            frame_info.append({'Video Name': video_name, 'Frame Path': frame_filename, 'Time in Seconds': current_time})
            extracted_times.add(current_time)
    cap.release()
    return frame_info

# Paths
video_directory = '/content/drive/MyDrive/work2/Test/Video'
key_positions_csv_path = '/content/drive/MyDrive/work2/Test/key_positions_result_2wav.csv'
output_base_folder = '/content/drive/MyDrive/work2/Test/Key_Frames'

# Create the output base folder if it doesn't exist
os.makedirs(output_base_folder, exist_ok=True)

# Load key positions from the CSV file
df = pd.read_csv(key_positions_csv_path)

# Process only the '2.mp4' video file
video_name = '2'
video_file_path = os.path.join(video_directory, f'{video_name}.mp4')

if not os.path.exists(video_file_path):
    print(f"Video file not found: {video_file_path}")
else:
    # Get key positions for '2.wav'
    row = df[df['File'] == '2.wav'].iloc[0]
    negative_positions = eval(row['negative_positions'])
    positive_positions = eval(row['positive_positions'])
    key_positions = list(set(negative_positions + positive_positions))

    # Create output folder for this video
    output_folder = os.path.join(output_base_folder, video_name)
    os.makedirs(output_folder, exist_ok=True)

    # Extract frames at key positions
    frame_info = extract_frames(video_file_path, key_positions, output_folder)

    # Save frame information to a new CSV file
    frames_csv_path = '/content/drive/MyDrive/work2/Test/Extracted_Key_Frames_Info_2mp4.csv'
    frame_info_df = pd.DataFrame(frame_info)
    frame_info_df.to_csv(frames_csv_path, index=False)

    print("Key frames extracted and saved for 2.mp4.")
    print(f"Frame information saved to {frames_csv_path}")
```

```python
import librosa
import numpy as np
import tensorflow as tf
import pandas as pd

# Load the dataset
audio_df = pd.read_csv('/content/drive/MyDrive/work2/Test/Frames_meta_data_2mp4.csv')

# Get the audio file name
audio_file = '2.wav'

# Load the pre-trained model
audio_model_path = '/content/drive/MyDrive/work2/Final/Audio_Models/Audio_Model_1_LSTM.h5'
audio_model = tf.keras.models.load_model(audio_model_path)

# Helper function for audio feature extraction
def extract_features(file_path):
    y, sr = librosa.load(file_path, duration=3)
    mfccs = librosa.feature.mfcc(y=y, sr=sr, n_mfcc=13)
    mel = librosa.feature.melspectrogram(y=y, sr=sr)
    contrast = librosa.feature.spectral_contrast(y=y, sr=sr)
    bandwidth = librosa.feature.spectral_bandwidth(y=y, sr=sr)
    energy_contour = librosa.feature.rms(y=y)
    spectral_rolloff = librosa.feature.spectral_rolloff(y=y, sr=sr)
    features = np.concatenate((
        np.mean(mfccs, axis=1),
        np.mean(mel, axis=1),
        np.mean(contrast, axis=1),
        np.mean(bandwidth, axis=1),
        np.mean(energy_contour, axis=1),
        np.mean(spectral_rolloff, axis=1)
    ))
    return features

# Predict audio sentiment
audio_path = f'/content/drive/MyDrive/work2/Test/Audio/{audio_file}'
audio_features = extract_features(audio_path)
audio_features_scaled = np.expand_dims(audio_features, axis=0)
audio_features_scaled = np.expand_dims(audio_features_scaled, -1)
audio_pred = audio_model.predict(audio_features_scaled)[0]
audio_sentiment = np.argmax(audio_pred)
audio_percentage = audio_pred[audio_sentiment]

# Save results to a DataFrame
results_df_audio1 = pd.DataFrame({
    'Modality': ['Audio Model 1'],
    'Sentiment': ['neutral' if audio_sentiment == 0 else 'positive' if audio_sentiment == 1 else 'negative'],
    'Percentage': [audio_percentage]
})

# Save to CSV
output_csv_path = '/content/drive/MyDrive/work2/Test/Temp_Results/Temp_results_audio1_2wav.csv'
results_df_audio1.to_csv(output_csv_path, index=False)
print(f"Results saved to {output_csv_path}")

# Print the results
print(f"Audio file: {audio_file}")
print(f"Sentiment: {results_df_audio1['Sentiment'][0]}")
print(f"Confidence: {results_df_audio1['Percentage'][0]:.2f}")
```

```python
import librosa
import numpy as np
import pandas as pd
import joblib
import os

# Load the pre-trained model
audio_model_path = '/content/drive/MyDrive/work2/Final/Audio_Models/Audio_Model_2_RandomForest.joblib'
audio_scaler_path = '/content/drive/MyDrive/work2/Final/Audio_Models/Audio_Model_2_RandomForest_Scaler.joblib'
audio_model = joblib.load(audio_model_path)
audio_scaler = joblib.load(audio_scaler_path)

# Helper function for audio feature extraction
def extract_features(file_path):
    y, sr = librosa.load(file_path, duration=3)
    mfccs = np.mean(librosa.feature.mfcc(y=y, sr=sr, n_mfcc=13), axis=1)
    chroma = np.mean(librosa.feature.chroma_stft(y=y, sr=sr), axis=1)
    mel = np.mean(librosa.feature.melspectrogram(y=y, sr=sr), axis=1)
    spectral_contrast = np.mean(librosa.feature.spectral_contrast(y=y, sr=sr), axis=1)
    tonnetz = np.mean(librosa.feature.tonnetz(y=librosa.effects.harmonic(y), sr=sr), axis=1)
    features = np.concatenate((mfccs, chroma, mel, spectral_contrast, tonnetz))
    return features

# Specify the audio file
audio_file = '2.wav'

# Predict audio sentiment
audio_path = f'/content/drive/MyDrive/work2/Test/Audio/{audio_file}'
audio_features = extract_features(audio_path)
audio_features_scaled = audio_scaler.transform([audio_features])
audio_pred = audio_model.predict_proba(audio_features_scaled)[0]
audio_sentiment = np.argmax(audio_pred)
audio_percentage = audio_pred[audio_sentiment]

# Save results to a DataFrame
results_df_audio2 = pd.DataFrame({
    'Modality': ['Audio Model 2'],
    'Sentiment': ['neutral' if audio_sentiment == 0 else 'positive' if audio_sentiment == 1 else 'negative'],
    'Percentage': [audio_percentage]
})

# Save to CSV
output_csv_path = '/content/drive/MyDrive/work2/Test/Temp_Results/Temp_results_audio2_2wav.csv'
results_df_audio2.to_csv(output_csv_path, index=False)
print(f"Results saved to {output_csv_path}")

# Print the results
print(f"Audio file: {audio_file}")
print(f"Sentiment: {results_df_audio2['Sentiment'][0]}")
print(f"Confidence: {results_df_audio2['Percentage'][0]:.2f}")
```

#Text Model 1 (BERT)

```python
import torch
from transformers import BertTokenizer, BertForSequenceClassification
from torch.utils.data import DataLoader, SequentialSampler, TensorDataset
import numpy as np
import pandas as pd

# Load the dataset
text_df = pd.read_csv('/content/drive/MyDrive/work2/Test/Frames_meta_data_2mp4.csv')

# Get the text transcription for audio 2
text = text_df[text_df['File'] == '2.wav']['Transcription'].iloc[0]

# Load the pre-trained model and tokenizer
tokenizer = BertTokenizer.from_pretrained('/content/drive/MyDrive/work2/Final/Text_Work/Text_model_1_BERT_Model')
model = BertForSequenceClassification.from_pretrained('/content/drive/MyDrive/work2/Final/Text_Work/Text_model_1_BERT_Model')

# Tokenize and encode the text
encoded_dict = tokenizer.encode_plus(
    text,
    add_special_tokens=True,
    max_length=64,
    truncation=True,
    padding='max_length',
    return_attention_mask=True,
    return_tensors='pt'
)

input_ids = encoded_dict['input_ids']
attention_mask = encoded_dict['attention_mask']

# Create DataLoader
dataset = TensorDataset(input_ids, attention_mask)
dataloader = DataLoader(dataset, sampler=SequentialSampler(dataset), batch_size=1)

# Predict text sentiment
model.eval()
with torch.no_grad():
    for batch in dataloader:
        inputs = {'input_ids': batch[0], 'attention_mask': batch[1]}
        outputs = model(**inputs)
        logits = outputs.logits
        preds = torch.softmax(logits, dim=1).cpu().numpy()

text_sentiment = np.argmax(preds)
text_percentage = preds[0][text_sentiment]

# Save results to a DataFrame
results_df_text1 = pd.DataFrame({
    'Modality': ['Text Model 1'],
    'Sentiment': ['neutral' if text_sentiment == 0 else 'positive' if text_sentiment == 1 else 'negative'],
    'Percentage': [text_percentage]
})

# Save to CSV
output_csv_path = '/content/drive/MyDrive/work2/Test/Temp_Results/Temp_results_text1_2wav.csv'
results_df_text1.to_csv(output_csv_path, index=False)
print(f"Results saved to {output_csv_path}")

# Print the results
print(f"Audio file: 2.wav")
print(f"Sentiment: {results_df_text1['Sentiment'][0]}")
print(f"Confidence: {results_df_text1['Percentage'][0]:.2f}")
```

#Video Model 1 (MobileNetV2)

```python
import os
import numpy as np
import pandas as pd
import cv2
import tensorflow as tf
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input

# Load the dataset
frames_df = pd.read_csv('/content/drive/MyDrive/work2/Test/Updated_Key_Frames_Info_2wav.csv')
video_folder_path = '/content/drive/MyDrive/work2/Test/Video'

# Specify the video file
video_file = '2.mp4'
video_name = '2'

# Helper function for video feature extraction
def preprocess_image(filepath):
    img = cv2.imread(filepath)
    if img is None:
        print(f"Failed to read image: {filepath}")
        return None
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    img = cv2.resize(img, (224, 224))
    img = preprocess_input(img)
    return img

# Load the pre-trained model
video_model_path = '/content/drive/MyDrive/work2/Final/Image_Models/Video_Model_1_CNN_with_MobileNetV2.h5'
video_model = tf.keras.models.load_model(video_model_path)

# Get the frames corresponding to the current video and key frames only
video_frames = frames_df[(frames_df['Video Name'].astype(str) == video_name) & (frames_df['Key_Frame'] == 'Y')]

predictions = []
for frame_path in video_frames['Frame Path']:
    frame_full_path = os.path.join('/content/drive/MyDrive/work2/Test', frame_path)  # Adjust this path if needed
    video_image = preprocess_image(frame_full_path)
    if video_image is not None:
        video_image = np.expand_dims(video_image, axis=0)
        video_pred = video_model.predict(video_image)[0]
        predictions.append(video_pred)

if predictions:
    # Average predictions
    average_pred = np.mean(predictions, axis=0)
    video_sentiment = np.argmax(average_pred)
    video_percentage = average_pred[video_sentiment]

    result = {
        'Modality': 'Video Model 1',
        'Sentiment': 'neutral' if video_sentiment == 0 else 'positive' if video_sentiment == 1 else 'negative',
        'Percentage': video_percentage
    }
else:
    result = {
        'Modality': 'Video Model 1',
        'Sentiment': 'N/A',
        'Percentage': 0.0
    }

# Save results to a DataFrame
results_df = pd.DataFrame([result])

# Save to CSV
output_csv_path = '/content/drive/MyDrive/work2/Test/Temp_Results/Temp_results_video1_2mp4.csv'
results_df.to_csv(output_csv_path, index=False)
print(f"Results saved to {output_csv_path}")

# Print the results
print(f"Video file: {video_file}")
print(f"Sentiment: {result['Sentiment']}")
print(f"Confidence: {result['Percentage']:.2f}")
```

```python
import os
import pandas as pd

# Load individual model results for video 2
results_audio1 = pd.read_csv('/content/drive/MyDrive/work2/Test/Temp_Results/Temp_results_audio1_2wav.csv')
results_audio2 = pd.read_csv('/content/drive/MyDrive/work2/Test/Temp_Results/Temp_results_audio2_2wav.csv')
results_text1 = pd.read_csv('/content/drive/MyDrive/work2/Test/Temp_Results/Temp_results_text1_2wav.csv')
results_text2 = pd.read_csv('/content/drive/MyDrive/work2/Test/Temp_Results/Temp_results_text2_2wav.csv')
results_video1 = pd.read_csv('/content/drive/MyDrive/work2/Test/Temp_Results/Temp_results_video1_2mp4.csv')
results_video2 = pd.read_csv('/content/drive/MyDrive/work2/Test/Temp_Results/Temp_results_video2_2mp4.csv')

# Define the video file
video_file = '2.mp4'
video_name = '2'

# Manually set the sentiment for the video (assuming sentiment '2' for demonstration)
actual_senti = '2'

# Define weights for each model
weights = {
    'Audio Model 1': 0.529,
    'Audio Model 2': 0.521,
    'Text Model 1': 0.990,
    'Text Model 2': 0.986,
    'Video Model 1': 0.643,
    'Video Model 2': 0.653
}

# Phase 1: Fusion within each modality
def fuse_modality_pair(model1, model2, weight1, weight2):
    combined = pd.concat([model1, model2], ignore_index=True)
    combined['Weighted Percentage'] = combined.apply(lambda row: row['Percentage'] * (weight1 if row['Modality'] == model1['Modality'].iloc[0] else weight2), axis=1)
    return combined.loc[combined['Weighted Percentage'].idxmax()]

audio_result = fuse_modality_pair(results_audio1, results_audio2, weights['Audio Model 1'], weights['Audio Model 2'])
text_result = fuse_modality_pair(results_text1, results_text2, weights['Text Model 1'], weights['Text Model 2'])
video_result = fuse_modality_pair(results_video1, results_video2, weights['Video Model 1'], weights['Video Model 2'])

# Combine results from all modalities
phase1_results = pd.DataFrame([audio_result, text_result, video_result])

print("Phase 1 Results (Fusion within each modality):")
print(phase1_results[['Modality', 'Sentiment', 'Percentage', 'Weighted Percentage']])

# Phase 2: Fusion across modalities
max_predicted_sentiment = phase1_results.loc[phase1_results['Percentage'].idxmax()]
max_sentiment = max_predicted_sentiment['Sentiment']
max_percentage = max_predicted_sentiment['Percentage']
max_weight = 0.33  # Adjust this weight as needed

# Calculate Final Weighted Percentage
phase1_results['Final Weighted Percentage'] = phase1_results.apply(
    lambda row: row['Weighted Percentage'] + (max_weight * max_percentage if str(row['Sentiment']) == str(max_sentiment) else 0),
    axis=1
)

# Determine the final sentiment
final_sentiment = phase1_results.loc[phase1_results['Final Weighted Percentage'].idxmax()]

print("\nFinal Ensemble Sentiment Result:")
print(final_sentiment)

# Save the final result to a CSV file
final_result_df = pd.DataFrame({
    'File': [video_file],
    'Actual Senti': [actual_senti],
    'Predicted Sentiment': [final_sentiment['Sentiment']],
    'Weighted Percentage': [final_sentiment['Final Weighted Percentage']],
    'Modality': [final_sentiment['Modality']],
    'Percentage': [final_sentiment['Percentage']]
})

# Define the path to the CSV file
csv_file_path = '/content/drive/MyDrive/work2/Test/Final_Evaluation_Ensemble_Result_2mp4.csv'

# Save the result
final_result_df.to_csv(csv_file_path, index=False)

print(f"Final ensemble result for {video_file} saved to {csv_file_path}")
```
